```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from shared import colors, init_plt
```

```python
init_plt()
df = pd.read_csv('../../data/accuracy.csv')
```

```python
ax: plt.Axes
fig, axs = plt.subplots(nrows=2, ncols=1, figsize=(7, 4), dpi=200, layout='constrained')
fig.suptitle('Occurrence distribution of execution time', fontsize=16, y=1.075)
fig.supylabel('Occurrences', fontsize=12)
fig.supxlabel('Execution time (seconds)', fontsize=12)

for ax in axs:
    # y axis
    ax.set_yticks([])

    # x axis
    ax.set_xlim(104.5, 111.5)
    ax.grid(axis='x', which='major', visible=True, linestyle='--', color='grey', alpha=0.4)

axs[0].set_title("Persistent Disk", fontsize=12)
axs[0].hist(
    df[df['disk_type'] == 'pd']['value'] / 1e3,
    bins=np.arange(105, 111, 0.1),
    color=colors['pd']
)

axs[1].set_title("Hyperdisk", fontsize=12)
axs[1].hist(
    df[df['disk_type'] == 'hd']['value'] / 1e3,
    bins=np.arange(105, 111, 0.015),
    color=colors['hd']
)

# axs[0]
plt.show()
```
